### Fashion Design Generator

The `generate_fashion_design` function takes several parameters like `gender`, `occasion`, `style`, `garment`, `color`, and `fabric` to create a detailed fashion design concept. It uses the `gemini-3.6-flash` model to generate the design details, including a design name, concept, silhouette, fabric, color, embellishments, accessories, footwear, styling, and an image prompt.

```python
import gradio as gr


def generate_design(
    gender,
    occasion,
    style,
    garment,
    color,
    fabric
):

    return generate_fashion_design(
        gender,
        occasion,
        style,
        garment,
        color,
        fabric
    )


app = gr.Interface(
    fn=generate_design,

    inputs=[
        gr.Dropdown(
            ["Women", "Men", "Unisex"],
            value="Women",
            label="Gender"
        ),

        gr.Dropdown(
            ["Wedding", "Party", "Casual", "Formal", "Festival"],
            value="Wedding",
            label="Occasion"
        ),

        gr.Dropdown(
            ["Modern Indian", "Traditional", "Modern", "Luxury"],
            value="Modern Indian",
            label="Fashion Style"
        ),

        gr.Dropdown(
            ["Lehenga", "Saree", "Kurta", "Anarkali", "Dress", "Suit"],
            value="Lehenga",
            label="Garment"
        ),

        gr.Textbox(
            value="Emerald Green",
            label="Color"
        ),

        gr.Dropdown(
            ["Silk", "Cotton", "Velvet", "Chiffon", "Organza"],
            value="Silk",
            label="Fabric"
        )
    ],

    outputs=gr.Markdown(),

    title="👗 AI Fashion Design Generator",

    description="Generate creative fashion designs using Google Gemini AI."
)

app.launch(share=True)
```

```python
import gradio as gr


def generate_design(gender, occasion, style, garment, color, fabric):

    result = generate_fashion_design(
        gender,
        occasion,
        style,
        garment,
        color,
        fabric
    )

    return result


demo = gr.Interface(
    fn=generate_design,

    inputs=[
        gr.Dropdown(
            ["Women", "Men", "Unisex"],
            value="Women",
            label="Gender"
        ),

        gr.Dropdown(
            ["Wedding", "Party", "Casual", "Formal", "Festival"],
            value="Wedding",
            label="Occasion"
        ),

        gr.Dropdown(
            ["Modern Indian", "Traditional", "Modern", "Luxury"],
            value="Modern Indian",
            label="Style"
        ),

        gr.Dropdown(
            ["Lehenga", "Saree", "Kurta", "Anarkali", "Dress", "Suit"],
            value="Lehenga",
            label="Garment"
        ),

        gr.Textbox(
            value="Emerald Green",
            label="Color"
        ),

        gr.Dropdown(
            ["Silk", "Cotton", "Velvet", "Chiffon", "Organza"],
            value="Silk",
            label="Fabric"
        )
    ],

    outputs=gr.Textbox(
        label="✨ Generated Fashion Design",
        lines=20
    ),

    title="👗 AI Fashion Design Generator",

    description="Enter your requirements and generate an AI fashion design."
)

demo.launch(share=True)
```
